```python
import pandas as pd
from google.colab import drive
import os
import numpy as np

# Drive mount
drive.mount('/content/drive', force_remount=True)

tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]
INPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook2_Clipped/'
OUTPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook3_FeatureEngineered/'

if not os.path.exists(OUTPUT_PATH):
    os.makedirs(OUTPUT_PATH)

for t in tickers:
    df = pd.read_csv(os.path.join(INPUT_PATH, f"{t}_clipped.csv"))

    # Tarih formatı
    df['Date'] = pd.to_datetime(df['FiscalDateEnding']).dt.strftime('%Y-%m-%d')

    # Feature Engineering: Growth ve oran sütunları
    # EPS Growth
    df['EPS_Growth'] = df['EPS'].pct_change()
    # Revenue Growth
    if 'Revenue' in df.columns:
        df['Revenue_Growth'] = df['Revenue'].pct_change()
    # FreeCashFlow / Revenue
    if 'FreeCashFlow' in df.columns and 'Revenue' in df.columns:
        df['FCF_to_Revenue'] = df['FreeCashFlow'] / df['Revenue']
    # Debt/Equity
    if 'TotalDebt' in df.columns and 'TotalEquity' in df.columns:
        df['Debt_to_Equity'] = df['TotalDebt'] / df['TotalEquity']
    # NetMargin
    if 'NetIncome' in df.columns and 'Revenue' in df.columns:
        df['NetMargin'] = df['NetIncome'] / df['Revenue']

    # Boş değerleri tekrar doldur (growth için oluşabilecek NaN)
    df.fillna(method='ffill', inplace=True)
    df.fillna(method='bfill', inplace=True)

    # Kaydet
    out_file = os.path.join(OUTPUT_PATH, f"{t}_feature_engineered.csv")
    df.to_csv(out_file, index=False)
    print(f"{t} için Feature Engineering tamamlandı: {out_file}")
```
