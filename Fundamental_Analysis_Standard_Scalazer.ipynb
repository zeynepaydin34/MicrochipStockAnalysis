```python
# Notebook 2: Aykırı değer kırpma (3-sigma)

import pandas as pd
from google.colab import drive
import os
import numpy as np

drive.mount('/content/drive', force_remount=True)

tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]
INPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook1_Filled/'
OUTPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook2_Clipped/'

if not os.path.exists(OUTPUT_PATH):
    os.makedirs(OUTPUT_PATH)

sigma_threshold = 3

for t in tickers:
    df = pd.read_csv(os.path.join(INPUT_PATH, f"{t}_filled.csv"))
    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    for col in numeric_cols:
        mean = df[col].mean()
        std = df[col].std()
        df[col] = df[col].clip(lower=mean-sigma_threshold*std, upper=mean+sigma_threshold*std)

    out_file = os.path.join(OUTPUT_PATH, f"{t}_clipped.csv")
    df.to_csv(out_file, index=False)
    print(f"{t} için aykırı değerler kırpıldı: {out_file}")
```
