```python
import pandas as pd
import numpy as np
from google.colab import drive

# -------------------------------
# 1) Google Drive'ı bağla
# -------------------------------
drive.mount('/content/drive', force_remount=True)

# -------------------------------
# 2) Hisseler ve klasör yolu
# -------------------------------
hisseler = ['AMD', 'NVDA', 'ARM', 'INTC','MU']
ANA_KLASOR = '/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/'

# -------------------------------
# 3) Feature Engineering Fonksiyonu
# -------------------------------
def add_features(df):
    windows = [10, 20, 50]

    # --- Rolling Mean & Std (Adj Close)
    for w in windows:
        df[f'MA_{w}'] = df['Adj Close'].rolling(window=w).mean()
        df[f'STD_{w}'] = df['Adj Close'].rolling(window=w).std()

    # --- Momentum (price)
    for w in [10, 20]:
        df[f'Momentum_{w}'] = df['Adj Close'] / df['Adj Close'].shift(w) - 1

    return df

# -------------------------------
# 4) Hisseleri döngü ile işleme
# -------------------------------
for h in hisseler:
    print(f"\n--- {h} için Feature Engineering başlıyor ---")

    input_path = f'{ANA_KLASOR}{h}_technical_final_clean.csv'
    output_path = f'{ANA_KLASOR}{h}_features_engineered.csv'

    # CSV yükleme
    try:
        df = pd.read_csv(input_path, index_col=0, parse_dates=True)
        df.index.name = "Date"
        print(f"✅ {h} yüklendi. Satır sayısı: {len(df)}")
    except Exception as e:
        print(f"❌ {h} yüklenirken hata: {e}")
        continue

    # Feature ekle
    df = add_features(df)

    # İlk satırlardaki NaN değerleri düşür (rolling ve lag nedeniyle oluşur)
    df.dropna(inplace=True)
    print(f"✅ Feature engineering tamamlandı. Satır sayısı: {len(df)}")

    # CSV kaydet
    df.to_csv(output_path, index=True)
    print(f"🎉 {h} için kaydedildi: {output_path}")
```
