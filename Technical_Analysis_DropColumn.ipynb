```python
import pandas as pd
from google.colab import drive

# Drive'ı bağla
try:
    drive.mount('/content/drive')
except Exception as e:
    print(f"Drive bağlantı hatası: {e}")

# CSV dosya yolu
input_file_path = '/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/AMD_technical_full.csv'
output_file_path = '/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/AMD_technical_filtered.csv'

# CSV'yi oku
df = pd.read_csv(input_file_path, index_col='Date')

# 'Date' indeksini datetime formatına çevir
df.index = pd.to_datetime(df.index)

# Başlangıç tarihi
start_date = '2000-01-11'

# Belirlenen tarihten itibaren veri setini filtrele
df_filtered = df.loc[df.index >= start_date].copy()

# Filtrelenmiş veri CSV olarak kaydedilsin
df_filtered.to_csv(output_file_path, index=True)

print(f"✅ Temizlenmiş ve filtrelenmiş veri seti kaydedildi: {output_file_path}")
```
