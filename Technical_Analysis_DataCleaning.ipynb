```python
import pandas as pd
from google.colab import drive
import numpy as np

# --- 1. Google Drive'ı bağla ---
drive.mount('/content/drive', force_remount=True)

# --- 2. Dosya yolları ve hisse sembolü ---
HISSE_SEMBOLU = 'AMD'
# INPUT olarak, ilk temizliği yaptığınız dosyayı kullanmalıyız.
INPUT_PATH = f'/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/{HISSE_SEMBOLU}_technical_filtered.csv'

print(f"\n--- AMD Veri Kontrol ve Raporlama İşlemi Başlıyor ---")

# --- 3. CSV Yükleme ve Datetime Çevirme (İstenen 1) ---
try:
    # 'Date' sütununu indeks olarak yükle ve datetime'a çevir (parse_dates=True)
    df = pd.read_csv(INPUT_PATH, index_col='Date', parse_dates=True)
    print(f"✅ {HISSE_SEMBOLU} verisi yüklendi ve 'Date' indeksi datetime'a çevrildi.")
except Exception as e:
    print(f"❌ HATA: Yükleme veya Datetime Çevirme Hatası: {e}")
    raise

# --- 4. Symbol sütununu başa al (İstenen 3) ---
if 'Symbol' not in df.columns:
    df['Symbol'] = HISSE_SEMBOLU
    print("✅ 'Symbol' sütunu eklendi.")
else:
    print("✅ 'Symbol' sütunu zaten mevcut.")

cols = ['Symbol'] + [col for col in df.columns if col != 'Symbol']
df_control = df[cols].copy()
print("✅ 'Symbol' sütunu başa alındı.")

# --- 5. Eksik Değer (NaN) Kontrolü (İstenen 2) ---
nan_counts = df_control.isnull().sum()
total_nan = nan_counts.sum()
print("\n--- 1. Eksik Değer (NaN) Kontrolü (Müdahalesiz) ---")
print(f"Toplam zaten mevcut NaN sayısı: {total_nan}")
if total_nan > 0:
    print("❌ TESPİT: NaN değerler mevcut. Sütun bazında dağılım:")
    print(nan_counts[nan_counts > 0])
else:
    print("✅ Toplam mevcut NaN sayısı 0. Veri temiz.")


# --- 6. Eksik İş Günü Kontrolü (Tatil Kontrolü) (İstenen 4) ---
df_control.sort_index(inplace=True)
full_dates = pd.date_range(start=df_control.index.min(), end=df_control.index.max(), freq='B')
df_reindexed = df_control.reindex(full_dates)

nan_after_reindex = df_reindexed.isnull().sum().sum()
# Bir satırda tek bir NaN bile olsa, o satırı eksik iş günü sayarız.
rows_with_nan = df_reindexed[df_reindexed.isnull().any(axis=1)]
rows_to_drop_count = len(rows_with_nan)

print("\n--- 2. Eksik İş Günü Kontrolü (Doldurmasız) ---")
if rows_to_drop_count > 0:
    print(f"❌ TESPİT: {rows_to_drop_count} adet eksik iş günü satırı bulundu.")
    print(f"Bu hisse Model A'ya girmeden önce doldurulmalıdır/atılmalıdır.")
    # Sütun bazında NaN sayısını göster (Her sütunda aynı çıkacaktır)
    nan_counts_reindex = df_reindexed.isnull().sum()
    print("Sütun bazında eklenen NaN sayısı:")
    print(nan_counts_reindex[nan_counts_reindex > 0])
else:
    print("✅ Eksik iş günü bulunamadı.")


# --- 7. Aykırı Değer (Outlier) Kontrolü (Sadece Kontrol) (İstenen 5) ---
# Eksik iş günlerini eklediğimiz tablo üzerinden yapalım (ffill yapmadığımız için daha doğru)
numeric_cols = df_reindexed.select_dtypes(include=np.number)
z_scores = ((numeric_cols - numeric_cols.mean()) / numeric_cols.std()).abs()
outliers_count = (z_scores > 3).sum()

print("\n--- 3. Aykırı Değer (3-Sigma) Kontrolü (Müdahalesiz) ---")
print(outliers_count[outliers_count > 0])
print(f"Toplam 3-Sigma dışı değer sayısı: {outliers_count.sum()}")
print("\n--- Raporlama Tamamlandı. Hiçbir müdahale/kaydetme yapılmadı. ---")
```

```python
import pandas as pd
from google.colab import drive
import numpy as np

# --- 1. Google Drive'ı bağla ---
drive.mount('/content/drive', force_remount=True)

# --- 2. Dosya yolları ve hisse sembolü ---
HISSE_SEMBOLU = 'AMD'
INPUT_PATH = f'/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/{HISSE_SEMBOLU}_technical_filtered.csv'
OUTPUT_PATH = f'/content/drive/MyDrive/ChipStockAnalysis/TEKNİK ANALİZ VERİLERİ/{HISSE_SEMBOLU}_technical_final_clean.csv'

print(f"\n--- AMD Nihai Temizlik İşlemi BAŞLIYOR (FFILL ile Doldurma ve Kaydetme) ---")

# --- 3. CSV Yükleme ve Hazırlık ---
try:
    df = pd.read_csv(INPUT_PATH, index_col='Date', parse_dates=True)
    df['Symbol'] = HISSE_SEMBOLU
    cols = ['Symbol'] + [col for col in df.columns if col != 'Symbol']
    df_clean = df[cols].copy()
    print(f"✅ {HISSE_SEMBOLU} verisi yüklendi.")
except Exception as e:
    print(f"❌ HATA: Yükleme hatası: {e}")
    raise

# -------------------------------------------------------------------
# EKSİK İŞ GÜNLERİNİ DOLDURMA (ffill)
# -------------------------------------------------------------------

df_clean.sort_index(inplace=True)
full_dates = pd.date_range(start=df_clean.index.min(), end=df_clean.index.max(), freq='B')
df_reindexed = df_clean.reindex(full_dates)

nan_count_before = df_reindexed.isnull().sum().sum()

if nan_count_before > 0:
    print(f"❌ UYARI: {nan_count_before} adet eksik hücre (242 eksik gün) FFILL ile dolduruluyor.")

    # AKSİYON: Tatil günlerini ffill ile doldurma (FFILL/BFILL YÖNTEMİ)
    df_reindexed = df_reindexed.ffill().bfill()

    if df_reindexed.isnull().sum().sum() == 0:
        print("✅ Eksik iş günleri başarıyla dolduruldu. Toplam NaN: 0")
    else:
        print("❌ HATA: Doldurmaya rağmen NaN kaldı!")
else:
    print("✅ Eksik iş günü bulunamadı.")

df_clean = df_reindexed.copy()

# --- 4. Aykırı Değer Kontrolü (Sadece Raporlama için Tekrar) ---
numeric_cols = df_clean.select_dtypes(include=np.number)
z_scores = ((numeric_cols - numeric_cols.mean()) / numeric_cols.std()).abs()
outliers_count = (z_scores > 3).sum()

print("\n--- Aykırı Değer Kontrolü ---")
print(f"Toplam 3-Sigma dışı değer sayısı: {outliers_count.sum()} (Değiştirilmedi)")

# --- 5. Kaydetme ---
df_clean.to_csv(OUTPUT_PATH, index=True)
print(f"\n🎉 AMD nihai temizliği tamamlandı ve kaydedildi: {OUTPUT_PATH}")
```
