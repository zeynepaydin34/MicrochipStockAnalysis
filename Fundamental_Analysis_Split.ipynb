```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from google.colab import drive
import os
import joblib

drive.mount('/content/drive', force_remount=True)

tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]
INPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook3_FeatureEngineered/'
OUTPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook4_Scaled/'

if not os.path.exists(OUTPUT_PATH):
    os.makedirs(OUTPUT_PATH)

split_train = 0.7
split_val = 0.15
split_test = 0.15

for t in tickers:
    df = pd.read_csv(os.path.join(INPUT_PATH, f"{t}_feature_engineered.csv"))

    df.sort_values('Date', inplace=True)

    target_col = 'FreeCashFlow'
    # Sadece sayısal sütunları al
    feature_cols = df.select_dtypes(include=np.number).columns.tolist()
    if target_col in feature_cols:
        feature_cols.remove(target_col)

    n = len(df)
    train_end = int(n*split_train)
    val_end = train_end + int(n*split_val)

    train_df = df.iloc[:train_end].copy()
    val_df = df.iloc[train_end:val_end].copy()
    test_df = df.iloc[val_end:].copy()

    # Scaling
    scaler_X = StandardScaler()
    train_df[feature_cols] = scaler_X.fit_transform(train_df[feature_cols])
    val_df[feature_cols] = scaler_X.transform(val_df[feature_cols])
    test_df[feature_cols] = scaler_X.transform(test_df[feature_cols])

    scaler_y = StandardScaler()
    train_df[target_col] = scaler_y.fit_transform(train_df[[target_col]])
    val_df[target_col] = scaler_y.transform(val_df[[target_col]])
    test_df[target_col] = scaler_y.transform(test_df[[target_col]])

    # Kaydet
    train_df.to_csv(os.path.join(OUTPUT_PATH, f"{t}_train_scaled.csv"), index=False)
    val_df.to_csv(os.path.join(OUTPUT_PATH, f"{t}_val_scaled.csv"), index=False)
    test_df.to_csv(os.path.join(OUTPUT_PATH, f"{t}_test_scaled.csv"), index=False)

    joblib.dump(scaler_X, os.path.join(OUTPUT_PATH, f"{t}_X_scaler.pkl"))
    joblib.dump(scaler_y, os.path.join(OUTPUT_PATH, f"{t}_y_scaler.pkl"))

    print(f"{t} için train/val/test split ve scaling tamamlandı.")
```
