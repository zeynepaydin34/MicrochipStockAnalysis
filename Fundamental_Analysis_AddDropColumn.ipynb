```python
import pandas as pd
from google.colab import drive
import os

# Drive'ı mount et
drive.mount('/content/drive')

# Hisseler listesi
tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]

# Klasör yolu (örnek)
base_path = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL ANALIZ VERİLERİ'

for t in tickers:
    try:
        # Temel CSV
        fundamentals_file = os.path.join(base_path, f"{t}_fundamentals_full.csv")
        df = pd.read_csv(fundamentals_file)

        # v2 CSV (EPS dolu)
        v2_file = os.path.join(base_path, f"{t}-v2.csv")
        df_v2 = pd.read_csv(v2_file)

        # v2 den sözlük oluştur (Date -> EPS)
        eps_dict = pd.Series(df_v2.EPS.values,index=df_v2.Date).to_dict()

        # Boş EPS'leri doldur
        df['EPS'] = df.apply(lambda row: eps_dict.get(row['FiscalDateEnding'], row['EPS']), axis=1)

        # Kaydet (yeni dosya)
        out_file = os.path.join(base_path, f"{t}_fundamentals_full_filled.csv")
        df.to_csv(out_file, index=False)
        print(f"{t} için EPS değerleri dolduruldu ve kaydedildi: {out_file}")

    except FileNotFoundError as e:
        print(f"Hata: {e}")
    except Exception as e:
        print(f"{t} için hata oluştu: {e}")
```

```python
import pandas as pd
from google.colab import drive
import os

# Drive'ı mount et
drive.mount('/content/drive')

# Hisseler listesi
tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]

# Klasör yolu
base_path = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL ANALIZ VERİLERİ'

for t in tickers:
    try:
        # EPS ve MarketCap doldurulmuş CSV
        filled_file = os.path.join(base_path, f"{t}_fundamentals_full_filled.csv")
        df = pd.read_csv(filled_file)

        # FreeCashFlow hesapla: mevcut değilse OperatingCashFlow - CapEX
        df['FreeCashFlow'] = df.apply(
            lambda row: row['OperatingCashFlow'] - row['CapEX']
            if pd.isna(row.get('FreeCashFlow')) else row['FreeCashFlow'],
            axis=1
        )

        # Kaydet (yeni dosya)
        out_file = os.path.join(base_path, f"{t}_fundamentals_full_filled_FCF.csv")
        df.to_csv(out_file, index=False)
        print(f"{t} için FreeCashFlow hesaplandı ve kaydedildi: {out_file}")

    except FileNotFoundError as e:
        print(f"Hata: {e}")
    except Exception as e:
        print(f"{t} için hata oluştu: {e}")
```

```python
import pandas as pd
from google.colab import drive
import os

# Drive'ı mount et
drive.mount('/content/drive')

# Hisseler listesi
tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]

# Klasör yolu
base_path = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL ANALIZ VERİLERİ'

# Silinecek sütunlar
drop_cols = ["PERatio","PBRatio","MarketCapitalization","PS_Ratio","EV","EV/EBITDA","Beta","DividendYield"]

for t in tickers:
    try:
        file_path = os.path.join(base_path, f"{t}_fundamentals_full_filled.csv")
        df = pd.read_csv(file_path)

        # Sütunları sil
        df = df.drop(columns=[col for col in drop_cols if col in df.columns])

        # Yeni dosya olarak kaydet
        out_file = os.path.join(base_path, f"{t}_fundamentals_cleaned.csv")
        df.to_csv(out_file, index=False)
        print(f"{t} için istenilen sütunlar silindi ve kaydedildi: {out_file}")

    except FileNotFoundError as e:
        print(f"Hata: {e}")
    except Exception as e:
        print(f"{t} için hata oluştu: {e}")
```

```python
import pandas as pd
from google.colab import drive
import os

drive.mount('/content/drive')

tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]
base_path = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL ANALIZ VERİLERİ'

for t in tickers:
    # Temizlenmiş dosyadan başla
    file_path = os.path.join(base_path, f"{t}_fundamentals_cleaned.csv")
    df = pd.read_csv(file_path)

    # Tarih kolonunu düzenle
    df['Date'] = pd.to_datetime(df['FiscalDateEnding']).dt.strftime('%Y-%m-%d')

    # FreeCashFlow hesapla
    df['FreeCashFlow'] = df['OperatingCashFlow'] - df['CapEX']

    # Kolon sıralaması
    cols = df.columns.tolist()
    if 'Symbol' in cols:
        cols.remove('Symbol')
        cols.remove('Date')
        df = df[['Symbol', 'Date'] + cols]
    else:
        df = df[['Date'] + cols]

    # Yeni, anlamlı dosya adı
    out_file = os.path.join(base_path, f"{t}_fundamentals_enriched.csv")
    df.to_csv(out_file, index=False)

    print(f"{t} için enriched dosya oluşturuldu: {out_file}")
```
