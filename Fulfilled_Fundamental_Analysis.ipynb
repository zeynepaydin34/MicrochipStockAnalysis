```python
# Notebook 1: Eksik değer doldurma (FFILL/BFILL) – enriched dosyadan başlıyoruz

import pandas as pd
from google.colab import drive
import os
import numpy as np

drive.mount('/content/drive', force_remount=True)

tickers = ["NVDA", "AMD", "ARM", "INTC", "MU"]
INPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL ANALIZ VERİLERİ'
OUTPUT_PATH = '/content/drive/MyDrive/ChipStockAnalysis/TEMEL_MODEL_VERILERİ/Notebook1_Filled/'

if not os.path.exists(OUTPUT_PATH):
    os.makedirs(OUTPUT_PATH)

for t in tickers:
    df = pd.read_csv(os.path.join(INPUT_PATH, f"{t}_fundamentals_enriched.csv"))

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    # Eksik değerleri doldur
    df[numeric_cols] = df[numeric_cols].ffill().bfill()

    # Kaydet
    out_file = os.path.join(OUTPUT_PATH, f"{t}_filled.csv")
    df.to_csv(out_file, index=False)
    print(f"{t} için eksik değerler dolduruldu ve kaydedildi: {out_file}")
```
